Задание 1

Определите, какому фильму было выставлено больше всего оценок 5.0.

```python
import pandas as pd
import os

# Проверяем, существуют ли файлы
if not os.path.isfile('/movies.csv') or not os.path.isfile('/ratings.csv'):
    print("Один или оба файла не найдены. Проверьте пути к файлам.")
else:
    # Загружаем данные
    movies = pd.read_csv('/movies.csv')
    ratings = pd.read_csv('/ratings.csv')

    # Фильтруем оценки 5.0
    five_star_ratings = ratings[ratings['rating'] == 5.0]

    # Подсчитываем количество оценок 5.0 для каждого фильма
    five_star_counts = five_star_ratings['movieId'].value_counts()

    # Получаем идентификатор фильма с максимальным количеством оценок 5.0
    top_movie_id = five_star_counts.idxmax()

    # Находим название фильма
    top_movie_title = movies[movies['movieId'] == top_movie_id]['title'].values[0]

    print(f'Фильм с максимальным количеством оценок 5.0: {top_movie_title}')
```

Задание 2

По данным файла power.csv посчитайте суммарное потребление стран Прибалтики (Латвия, Литва и Эстония) категорий 4, 12 и 21 за период с 2005 по 2010 год. Не учитывайте в расчётах отрицательные значения quantity.

```python
import pandas as pd

# Загружаем данные из файлов
power_df = pd.read_csv('/power.csv')

# Фильтруем данные
Baltic_countries = ['Latvia', 'Lithuania', 'Estonia']
categories = [4, 12, 21]
filtered_data = power_df[
    (power_df['country'].isin(Baltic_countries)) &
    (power_df['category'].isin(categories)) &
    (power_df['year'].between(2005, 2010)) &
    (power_df['quantity'] > 0)
]

# Суммируем потребления
total_consumption = filtered_data['quantity'].sum()

print(f"Суммарное потребление: {total_consumption}")
```
