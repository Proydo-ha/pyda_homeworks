Задание 1
Напишите функцию, которая классифицирует фильмы из материалов занятия по правилам:

оценка 2 и ниже — низкий рейтинг;
оценка 4 и ниже — средний рейтинг;
оценка 4.5 и 5 — высокий рейтинг.
Результат классификации запишите в столбец class.

```python
import pandas as pd

def classify_movies(movies_file, ratings_file):
    # Читаем данные из CSV файлов
    movies = pd.read_csv(movies_file)
    ratings = pd.read_csv(ratings_file)
    
    # Объединяем данные по movieId
    merged_data = pd.merge(movies, ratings[['movieId', 'rating']], on='movieId')

    # Функция для классификации рейтинга
    def classify(rating):
        if rating <= 2:
            return 'низкий рейтинг'
        elif rating <= 4:
            return 'средний рейтинг'
        elif rating <= 5:
            return 'высокий рейтинг'
    
    # Применяем функцию классификации и создаем новый столбец class
    merged_data['class'] = merged_data['rating'].apply(classify)

    # Возвращаем результат с классами
    return merged_data[['movieId', 'title', 'class']]

# Пример использования функции
result = classify_movies('movies.csv', 'ratings.csv')
print(result)
```

Используйте файл keywords.csv.

Нужно написать гео-классификатор, который каждой строке сможет выставить географическую принадлежность определённому региону. Т. е. если поисковый запрос содержит название города региона, то в столбце ‘region’ пишется название этого региона. Если поисковый запрос не содержит названия города, то ставим ‘undefined’.

Правила распределения по регионам Центр, Северо-Запад и Дальний Восток:

geo_data = {
1 'Центр': ['москва', 'тула', 'ярославль'],
2
3  'Северо-Запад': ['петербург', 'псков', 'мурманск'],
4
5 'Дальний Восток': ['владивосток', 'сахалин', 'хабаровск']
}

Результат классификации запишите в отдельный столбец region.

ОБРАТИТЕ ВНИМАНИЕ! 
Я не очень поняла суть файла keywords.csv, так как открыв его, я увидела кучу непонятных символов (например, РѕРґРЅРѕРєР»Р°СЃСЃРЅРёРєРё,63810309 - и в таком же роде почти все). В связи с этим, в конечном файле все строки с регионом 'undefined'.

```python
import pandas as pd

# Загрузка данных
df = pd.read_csv('keywords.csv')

# Определение географических данных
geo_data = {
    'Центр': ['москва', 'тула', 'ярославль'],
    'Северо-Запад': ['петербург', 'псков', 'мурманск'],
    'Дальний Восток': ['владивосток', 'сахалин', 'хабаровск']
}

def classify_region(keyword):
    for region, cities in geo_data.items():
        if any(city in keyword.lower() for city in cities):
            return region
    return 'undefined'

# Применение классификации
df['region'] = df['keyword'].apply(classify_region)

# Сохранение результата
df.to_csv('keywords_with_regions.csv', index=False)
```

Есть мнение, что раньше снимали настоящее кино, не то что сейчас. Ваша задача — проверить это утверждение, используя файлы с рейтингами фильмов из прошлого домашнего занятия: файл movies.csv и ratings.csv из базы. Нужно проверить, верно ли, что с ростом года выпуска фильма его средний рейтинг становится ниже.

Вы не будете затрагивать субьективные факторы выставления этих рейтингов, а пройдётесь по алгоритму:

В переменную years запишите список из всех годов с 1950 по 2010 года.

Напишите функцию production_year, которая каждой строке из названия фильма выставляет год выпуска. Не все названия фильмов содержат год выпуска в одинаковом формате, поэтому используйте алгоритм:

для каждой строки пройдите по всем годам списка years;
если номер года присутствует в названии фильма, то функция возвращает этот год, как год выпуска;
если ни один из номеров года списка years не встретился в названии фильма, то возвращается 1900 год.
Запишите год выпуска фильма по алгоритму пункта 2 в новый столбец ‘year’.

Посчитайте средний рейтинг всех фильмов для каждого значения столбца ‘year’ и отсортируйте результат по убыванию рейтинга.

```python
# создадим список лет
years = list(range(1950, 2011))

# функция для извлечения года
def production_year(title):
    for year in years:
        if str(year) in title:
            return year
    return 1900

# добавим новый столбец в DataFrame
import pandas as pd
    # загрузка файлов
movies = pd.read_csv('movies.csv')
movies['year'] = movies['title'].apply(production_year)

# расчет среднего рейтинга по годам
    # загрузка рейтинги
ratings = pd.read_csv('ratings.csv')
    # объединяем данные
merged_data = pd.merge(ratings, movies, on='movieId')
    # средний рейтинг по годам
average_rating = merged_data.groupby('year')['rating'].mean().reset_index()
    # сортировка по убыванию
average_rating = average_rating.sort_values(by='rating', ascending=False)

# создадим новую таблицу со средними рейтингами
average_rating_table = average_rating.sort_values(by='rating', ascending=False)

# проверка на наличие данных в таблице
if not average_rating_table.empty:
    # вывод успешного сообщения о создании новой таблицы
    print('Новая таблица со средними рейтингами по годам успешно создана!')
else:
    # вывод сообщения об ошибке создания таблицы, если таблица пуста
    print('Ошибка создания новой таблицы со средними рейтингами по годам!')

# сохраняем новую таблицу в файл
average_rating_table.to_csv('average_rating_by_year.csv', index=False)
```
